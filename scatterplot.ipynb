```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

top_weather = df['Weather_Condition'].value_counts().head(10).index

filtered_df = df[df['Weather_Condition'].isin(top_weather)]

heatmap_data = pd.crosstab(
    filtered_df['Weather_Condition'],
    filtered_df['Severity'],
    normalize='index'
) * 100

plt.figure(figsize=(10, 8))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt='.1f',
    cmap='YlOrRd',
    linewidths=0.5,
    cbar_kws={'label': 'Percentage of Crashes (%)'}
)

plt.title('Crash Severity Distribution by Weather Condition',
          fontsize=16, fontweight='bold')

plt.xlabel('Crash Severity')
plt.ylabel('Weather Condition')

plt.tight_layout()
plt.show()
```
